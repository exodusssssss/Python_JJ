```python
import matplotlib.pyplot as plt
import random

def traj_urne(a,b,n_final=200):
    boules_a = 0
    boules_b = 0
    
    proportions = [boules_a / (boules_a + boules_b)]
    
    for _ in range(n_final):
        total = boules_a + boules_b
        prob_a = boules_a / total
        
        # Tirage au sort : on génère un nombre entre 0 et 1
        if random.random() < prob_a:
            boules_a += 1  # On a tiré A, on ajoute une boule A
        else:
            boules_b += 1  # On a tiré B, on ajoute une boule B
            
        proportions.append(boules_a / (boules_a + boules_b))
        
    return proportions

plt.plot(traj_urne(1, 1, n_final=200))
```
